# Data Cleaning

```python
#checking missing values
df.isnull().sum()
```

```python
#Filling missing values
df["Arrival Delay in Minutes"] = df["Arrival Delay in Minutes"].fillna(
    df["Arrival Delay in Minutes"].mean()
)
```

# Data Visualization

```python
#Passenger Distribution by Gender
df["Gender"].value_counts().plot(kind="bar")
plt.title("Passenger Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
#Passenger Satisfaction
df["satisfaction"].value_counts().plot(kind="pie",autopct="%1.1f%%")
plt.title("Passenger Satisfaction")
plt.ylabel("")
plt.show()
```

```python
#Customer Type
df["Customer Type"].value_counts().plot(kind="bar")
plt.title("Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Count")
plt.show()
```

```python
#Travel Class
df["Class"].value_counts().plot(kind="bar")
plt.title("Travel Class")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
```

```python
#Flight Distance Distribution
df["Flight Distance"].plot(kind="hist",bins=20)
plt.title("Flight Distance Distribution")
plt.xlabel("Distance")
plt.show()
```

```python
#Departure Delay
df["Departure Delay in Minutes"].plot(kind="box")
plt.title("Departure Delay")
plt.show()
```

# Correlation Analysis

```python
numeric_df = df.select_dtypes(include=np.number)
corr = numeric_df.corr()
corr
```

```python
#visual
plt.figure(figsize=(12,8))

sns.heatmap(corr,
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Matrix")

plt.show()
```

# Customer Segmentation

```python
#Satisfaction by Gender
df.groupby("Gender")["Flight Distance"].mean()
```

```python
#Satisfaction by Customer Type
df.groupby("Customer Type")["Flight Distance"].mean()
```

```python
#Satisfaction by Travel Class
df.groupby("Class")["Flight Distance"].mean()
```

```python
#Delay by Travel Class
df.groupby("Class")[["Departure Delay in Minutes",
                     "Arrival Delay in Minutes"]].mean()
```
